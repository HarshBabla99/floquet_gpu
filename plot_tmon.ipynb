```python
from plot import *
from dynamiqs.plot import gifit
```

```python
ng_values = np.arange(0.0, 0.5, 0.02)

flq_plt = FloquetPlot(
    directory = "/home/hkb7/project_pi_sp979/hkb7/dust/tmon_gpu/zeta_0.4000/",
    zeta_value = 0.40,
    ng_values = ng_values,
    fontsize = 18,
)
```

```python
def plot_01_landscape(ng_idx: int, interpolation='bilinear'):
    fig, axs = plt.subplots(2, 1, figsize=(24,7), sharex=True, sharey=True, constrained_layout=True)
    _ = flq_plt.landscape(idx=ng_idx, state_idx=0, fig=fig, ax=axs[0], interpolation=interpolation)
    _ = flq_plt.landscape(idx=ng_idx, state_idx=1, fig=fig, ax=axs[1], interpolation=interpolation)

    axs[0].set_xlabel(None)

    _zeta = flq_plt.zeta_value
    _EJ_over_EC = 8/(_zeta**2)
    _ng = fr"{ng_values[ng_idx]:.3f}" if ng_idx is not None else r"\text{averaged}"

    fig.suptitle(fr'$\zeta = {_zeta} \, (E_J/E_C = {_EJ_over_EC:.2f}) \, , n_g = {_ng}$', fontsize=20)
```

```python
plot_01_landscape(0)
plt.show()
```

```python
plot_01_landscape(None, interpolation='none')
plt.show()
```
